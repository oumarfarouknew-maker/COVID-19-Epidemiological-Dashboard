```python
## 1. Case Fatality Rate (CFR)

Case Fatality Rate represents the proportion of confirmed COVID-19 cases that resulted in death.

Formula:

CFR = (Cumulative_deaths / Cumulative_cases) × 100
```

```python
df['CFR'] = (
    df['Cumulative_deaths'] /
    df['Cumulative_cases']
) * 100
```

```python
## 2. Mortality Rate

Mortality Rate measures the number of deaths caused by COVID-19 
relative to the total population.

Formula:

Mortality Rate = (Total Deaths / Population) × 100,000

This indicator shows the impact of COVID-19 on the entire population.
```

```python
## 3. Infection Rate

Infection Rate measures the proportion of the population that was infected 
with COVID-19.

Formula:

Infection Rate = (Cumulative_cases / Population) × 100

This indicator helps estimate the spread of the disease within a population.
```

```python
## Note

Infection Rate was not calculated because population data 
is not available in the current dataset.
```

```python
## 4. Growth Rate

Growth Rate measures the percentage change in new COVID-19 cases 
compared with the previous day.

Formula:

Growth Rate = ((Current Cases - Previous Cases) / Previous Cases) × 100

This indicator helps identify periods of rapid increase or decrease 
in disease transmission.
```
